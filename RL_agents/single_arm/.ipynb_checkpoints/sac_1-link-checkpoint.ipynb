```python
import random
from copy import deepcopy
import numpy as np
from collections import namedtuple, deque
import matplotlib.pyplot as plt
import time

#pytorch
import torch
import torch.nn as nn
import torch.optim as optima
import torch.nn.functional as F
from torch.distributions.normal import Normal

from task import SingleLink as one_link_arm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
## Memory Replay
Transition = namedtuple('Transition', ('state','next_state','action','reward','done'))
                        
class Memory(object):
    def __init__(self, capacity):
        self.memory = deque(maxlen=capacity)
        self.capacity = capacity

    def push(self, state, next_state, action, reward, done):
        self.memory.append(Transition(
            state, next_state, action, reward, done))    
    def sample(self, batch_size):

        transitions = random.sample(self.memory, batch_size)
        batch = Transition(*zip(*transitions))
        return batch
    
    def __len__(self):
        return len(self.memory)
```

```python
## Model
class Critic(nn.Module): #critic_model
    def __init__(self,num_states,num_actions,num_hidden_l1,num_hidden_l2):
        super(Critic, self).__init__()

        self.layer = nn.Sequential(
            nn.Linear(num_states+num_actions,num_hidden_l1),
            nn.ReLU(),
            #nn.Linear(num_hidden_l1,num_hidden_l2),
            #nn.ReLU(),
            #nn.Linear(num_hidden_l2,num_actions),
            #nn.Linear(num_hidden_l2,1),
            nn.Linear(num_hidden_l1,1),
            )
         
    def forward(self, state, action):
        x = self.layer( torch.cat([state,action],dim=-1) ) 
        return x

class Actor(nn.Module): #actor_model
    def __init__(self,num_states,num_actions,num_hidden_l1,num_hidden_l2,act_limit):
        super(Actor, self).__init__()
        
        self.net = nn.Sequential(
            nn.Linear(num_states,num_hidden_l1),
            nn.ReLU(),
            #nn.Linear(num_hidden_l1,num_hidden_l2),
            #nn.ReLU(),     
            )
        
        #self.mean = nn.Linear(num_hidden_l2,num_actions)
        #self.log_std = nn.Linear(num_hidden_l2,num_actions)
        self.mean = nn.Linear(num_hidden_l1,num_actions)
        self.log_std = nn.Linear(num_hidden_l1,num_actions)
        self.act_limit = act_limit

    def forward(self, state):
        x = self.net(state)
        mu = self.mean(x)
        log_std = self.log_std(x)

        N = 1.00
        log_std = torch.clamp(log_std,-20*N,2*N) #limit std(-20,2)
        std  = torch.exp(log_std)

        pi_distribution = Normal(mu,std)
        action_u = pi_distribution.rsample() #sample actions

        #log_likelihood
        log_mu = pi_distribution.log_prob(action_u)
        log_pi = log_mu - (2*(np.log(2) - action_u - F.softplus(-2*action_u))) 

        #actions
        action = torch.tanh(action_u) #bound action [-1,1]
        action = self.act_limit*action #scale actions

        return action, log_pi
```

```python
#state normalization
def normalization(obs):
    high = np.pi
    low = -np.pi
    obs = 2*((obs - low)/(high - low)) - 1
    return obs

## Policy 
def get_action(state):
    with torch.no_grad():
        action, _ = pi_net(state)
        return action[0].numpy()

## Immediate update
def update_target_model(net, target_net):
    target_net.load_state_dict(net.state_dict())
    target_net.eval()

## Polyak update
def smooth_update_target_model(net, target_net):
    rho = 1 - (1/tau) #decay rate 
    for p_tgt, p in zip(target_net.parameters(),net.parameters()):
        p_tgt.data.mul_(1.0 - rho)
        p_tgt.data.add_(rho*p.data) 
```

```python
#compute Q losses
def compute_loss_q(states,actions,next_states,rewards,dones):
    # print('states: ',states)
    # print('actions: ', actions)
    # print('next_states: ',next_states)
    # combined = torch.cat([states,actions], dim=-1)
    # print(combined.shape)
    # test = self.pi_net(next_states)
    # print('test passed')

    q1 = q1_net(states,actions)
    q2 = q1_net(states,actions)
    #print('q1: ',q1)

    with torch.no_grad():
        next_actions, next_log_pis = pi_net(next_states) #sample actions

        tgt_q1 = tgt_q1_net(next_states,next_actions)
        tgt_q2 = tgt_q2_net(next_states,next_actions)
        tgt_q = torch.min(tgt_q1,tgt_q2)

        y = rewards + gamma*(1.0 - dones)*(tgt_q - alpha*next_log_pis)
        #y = y.squeeze(0)
        #print('y: ', y)

    loss_q1 = ((q1-y)**2).mean() 
    loss_q2 = ((q2-y)**2).mean() 
    #print('loss_q1: ', loss_q1)

    return loss_q1, loss_q2
```

```python
#compute Policy losses
def compute_loss_pi(states):
    actions,log_pis = pi_net(states) #sample actions
    q1 = q1_net(states,actions)
    q2 = q2_net(states,actions)
    q = torch.min(q1,q2)

    mod_log_pis = alpha*log_pis.squeeze(0)

    #print('q_a: ', q)
    #print('a*log_pi: ', mod_log_pis)

    loss_pi = (q - mod_log_pis).mean()
    #print('loss_pi: ',loss_pi)

    return -loss_pi #-ve for gradient ascent
```

```python
## Train model
def train_model(batch, 
                pi_optim,
                q1_optim,
                q2_optim):

    states =  torch.stack(batch.state) #torch.cat(batch.state) #
    next_states = torch.stack(batch.next_state) #torch.cat(batch.next_state)
    #actions = torch.cat(batch.action) 
    actions = torch.tensor(batch.action).reshape(batch_size,-1)
    rewards = torch.tensor(batch.reward).reshape(batch_size,-1)
    rewards = rewards.float()
    dones = torch.tensor(batch.done).reshape(batch_size,-1)
    dones = dones.float()

    #print('next_states: ', next_states)
    #print('states: ', states)
    #print('actions: ', actions)
    #print('rewards: ', rewards)
    #print('dones: ', dones)

    #Q-functions updates
    q1_optim.zero_grad() #clear data
    q2_optim.zero_grad() #clear data

    loss_q1, loss_q2 = compute_loss_q(states,
                                      actions,
                                      next_states,
                                      rewards,
                                      dones)
    loss_q1.backward() #gradients
    loss_q2.backward() #gradients
    q1_optim.step() #update
    q2_optim.step() #update

    #freeze Q-networks to avoid computing during policy update
    for p in q1_net.parameters():
        p.requires_grad = False
    for p in q2_net.parameters():
        p.requires_grad = False 

    #policy updates
    pi_optim.zero_grad() #clear data
    loss_pi = compute_loss_pi(states)
    #print('loss_pi: ', loss_pi)

    loss_pi.backward() #gradients
    pi_optim.step #update

    #unfreeze Q-networks to compute on next iteration
    for p in q1_net.parameters():
        p.requires_grad = True
    for p in q2_net.parameters():
        p.requires_grad = True

    #polyak updating of target networks
    with torch.no_grad():
        smooth_update_target_model(q1_net, tgt_q1_net)
        smooth_update_target_model(q2_net, tgt_q2_net)

    return loss_pi.item(), loss_q1.item(), loss_q2.item()
```

```python
env = one_link_arm(head=1) #call env
start_time = time.time()
```

```python
## Configurations
#hyperameters
batch_size = 64
tau = 200
replay_memory_capacity = int(1e6)
memory = Memory(replay_memory_capacity)
initial_exploration = 500 
lr = 0.1 #3e-4 
num_episodes = 100 
steps = 0
print_every = 10
alpha = 0.1 #0.2
gamma = 0.98 #0.995
```

```python
#model 
num_states = env.observation_space.shape[0]
#print('num_states: ',num_states)
num_actions = env.action_space.shape[0]
#print('num_actions: ',num_actions)
num_hidden_l1 = 3 #256 
num_hidden_l2 = 256  
act_limit = env.action_space.high[0]
#print('act_limit: ',act_limit)

#declare model
pi_net = Actor(num_states,
              num_actions,
              num_hidden_l1,
              num_hidden_l2,
              act_limit)

q1_net = Critic(num_states,
                num_actions,
                num_hidden_l1,
                num_hidden_l2)

q2_net = Critic(num_states,
               num_actions,
               num_hidden_l1,
               num_hidden_l2)

#target networks
tgt_q1_net = deepcopy(q1_net)
tgt_q2_net = deepcopy(q1_net)

#freeze target networks to only update via polyak averaging
for p in tgt_q1_net.parameters():
    p.requires_grad = False
for p in tgt_q2_net.parameters():
    p.requires_grad = False           

#optimizer & loss function
pi_optim = optima.Adam(pi_net.parameters(), lr=lr)
q1_optim = optima.Adam(q1_net.parameters(), lr=lr)
q2_optim = optima.Adam(q2_net.parameters(), lr=lr)
```

```python
pi_losses = []
q1_losses = []
q2_losses = []
cum_rewards = []
cum_running_score = []
avg_pi_loss = []
avg_q1_loss = []
avg_q2_loss = []
```

```python
for episode in range(num_episodes):
    done = False
    #steps = 0 
    score = 0

    obs = env.reset()
    #print('obs: ', obs*(180/np.pi) )
    norm_obs = obs #self.normalization(obs)
    #print('norm_obs: ', norm_obs)

    state = torch.tensor(norm_obs).float() 
    #print('state: ', state)

    while not done:
        steps += 1
        action = get_action(state)
        #action = self.env.action_space.sample()[0]
        #print('action: ', action)

        next_obs, reward, done, info = env.step(action)
        #next_norm_obs = self.normalization(next_obs)
        next_state = torch.tensor(next_obs).float()  #torch.tensor(next_norm_obs).float() 
        #print('next_state: ',next_state)
        #print('raw_rewards: ',reward[0])
        #print('steps: ', steps)
        #print('done: ', done)

        mask = 0 
        score += reward

        if done:
            mask = 1

        action = torch.tensor(action).unsqueeze(0).float() #convert list to tensor
        memory.push(state, next_state, action, reward, mask)
        state = next_state   

        #time.sleep(.1)
    #"""    
        if steps > initial_exploration:
            #print('intial_pass') 
            batch = memory.sample(batch_size)
            pi_loss, q1_loss, q2_loss = train_model(batch, 
                                                         pi_optim,
                                                         q1_optim,
                                                         q2_optim,)
            pi_losses.append(pi_loss)
            q1_losses.append(q1_loss)
            q2_losses.append(q2_loss)

    if pi_losses: #execute if not empty
        avg_pi_loss.append( np.mean(pi_losses) )
        avg_q1_loss.append( np.mean(q1_losses) )
        avg_q2_loss.append( np.mean(q2_losses) )

    #score = score #if score == 500.0 else score #+ 1
    #print('score: ', score)
    cum_rewards.append(score)
    #print('self.cum_rewards: ',self.cum_rewards)
    avg_reward = np.mean(cum_rewards[-10:])
    #std_reward = np.std(self.cum_rewards[-self.window_size:])
    #self.cum_running_score.append(avg_reward)
    #self.std_running_score.append(std_reward)
    
    if episode % print_every == 0:
        print("Episode: {} | Score: {} | Avg_reward: {}".format(
                                                    episode,
                                                    score,
                                                    avg_reward))

    cum_running_score.append( avg_reward )

end_time = time.time()
duration = end_time - start_time
```
